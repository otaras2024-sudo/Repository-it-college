## Завдання 1. Створення першого класу


```python

class Student:
    def __init__(self, name, group):
        self.name = name
        self.group = group

    def show_info(self):
        return f"Студент: {self.name}, група: {self.group}"


student1 = Student("Oleksandr", "КН-31")
print(student1.show_info())
```

```python

class StudentInfo:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    @property
    def information(self):
        return f"Ім'я: {self.name}, Вік: {self.age}"


student2 = StudentInfo("Oleksandr", 17)
print(student2.information)
```

```python

class StudentMethods:
    total_students = 0

    def __init__(self, name):
        self.name = name
        StudentMethods.total_students += 1

    @staticmethod
    def say_hello():
        return "Hello, students!"

    @classmethod
    def get_total_students(cls):
        return cls.total_students


student3 = StudentMethods("Oleksandr")
student4 = StudentMethods("Marta")

print(StudentMethods.say_hello())
print("Total students:", StudentMethods.get_total_students())
```
